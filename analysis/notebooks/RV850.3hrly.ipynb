```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import warnings
```

```python
NAME = 'RV850'
LONGNAME = '850 hPa relative vorticity times'
MONTHS_NH = [6,7,8,9,10]  # NH TC season months [1: Jan - 12: Dec]
MONTHS_SH = [1,2,3,4,5]   # SH TC season months
```

```python
# compute relative vorticity with average winds
ds = xr.open_dataset(f'../jobs/{NAME}/data/{NAME}.nc')
ds[NAME] = ds[NAME] * np.sign(ds.lat) / 1e-5

# add anomalies to datasets
for exp in ['cnt-ref','sai-ref','sai-cnt']:
    exp1,exp2 = exp.split('-')
    diff = ds.sel(exp=exp1) - ds.sel(exp=exp2)
    diff = diff.assign_coords(exp=exp)
    ds = xr.concat((ds,diff), dim='exp')

# add month weights and gaussian latitude weights
ds['dpm'] = ('month', [31,28,31,30,31,30,31,31,30,31,30,31])
with xr.open_dataset('../data/gw.nc').gw as gw:
    ds['gw'] = gw

# add 2D latitude and longitude for plotting
ds['lat2d'], ds['lon2d'] = xr.broadcast(ds.lat, ds.lon)
ds = ds.set_coords(['lat2d', 'lon2d'])

# compute 4 degree running mean (cyclical in lon)
yw = 17
xw = 13
ds[NAME] = ds[NAME].pad(lon=xw//2, mode='wrap').rolling(lon=xw, lat=yw, min_periods=1).mean().isel(lon=slice(xw//2,-xw//2)).where(ds[NAME].notnull())

# print annual mean quantile information for plotting limits
data = ds[NAME].weighted(ds.dpm).mean('month')
print('exp','[min, 0.02 pct, 0.98 pct, max]')
for exp in data.exp.data:
    print(exp, data.sel(exp=exp).quantile([0,0.02,0.98,1]).data)
```

```python
cfkwargs_abs = dict(levels=np.arange(-2,2.1,1/3), vmin=-2.25, vmax=2.25, extend='both', cmap='PuOr_r', transform=ccrs.PlateCarree(), transform_first=True)
cfkwargs_rel = dict(levels=np.arange(-.4,.41,0.1), vmin=-.5, vmax=.5, extend='both', cmap='RdBu_r', transform=ccrs.PlateCarree(), transform_first=True)
cbarunit_abs = r'$10^{-5}$ s$^{-1}$'
cbarunit_rel = r'$10^{-5}$ s$^{-1}$'
cbticks_abs = range(-2,3)
```

```python
data = ds[NAME].weighted(ds.dpm).mean('month')


kwargs = dict(transform=ccrs.PlateCarree(), transform_first=True)

fig,axs = plt.subplots(1,3, dpi=300, figsize=(12,4), layout='constrained', subplot_kw={'projection':ccrs.PlateCarree(central_longitude=200)})

ax=axs[0]
pcm = ax.contourf(data.lon2d, data.lat2d, data.sel(exp='ref'), **cfkwargs_abs)
#psm = ax.contour(data.lon2d, data.lat2d, data.sel(exp='ref'), levels=pcm.levels, colors='k', linewidths=0.3, **kwargs)
#pcm.set_rasterized(True)
#psm.set_rasterized(True)
#ax.contour(data.lon2d, data.lat2d, data.sel(exp='ref'), levels=[25], colors='k', linewidths=1.2, linestyles='--', **kwargs)
cb = fig.colorbar(pcm, ax=axs[0], shrink=0.45, ticks=cbticks_abs, orientation='vertical')
cb.ax.set_ylabel(cbarunit_abs)
ax.set_title('Reference')

for ax,exp,tit in zip(axs[1:],['cnt-ref','sai-ref'],['Control $-$ Ref.','SAI $-$ Ref.']):
    pcm = ax.contourf(data.lon2d, data.lat2d, data.sel(exp=exp), **cfkwargs_rel)
    #psm = ax.contour(data.lon2d, data.lat2d, data.sel(exp=exp), levels=pcm.levels, colors='k', linewidths=[0.3], **kwargs)
    #pcm.set_rasterized(True)
    ax.set_title(tit)
cb = fig.colorbar(pcm, ax=axs[1:], shrink=0.45, orientation='vertical')
cb.ax.set_ylabel(cbarunit_rel)

for ax in axs.flatten():
    ax.gridlines()
    #ax.coastlines()
    ax.add_feature(cfeature.LAND, edgecolor='k', zorder=2)

fig.suptitle('annual mean '+LONGNAME, y=0.9)
fig.savefig(f'../figures/{NAME}.annualmean.pdf', bbox_inches='tight', transparent=True)
```

#### Effects of horizontal averaging (make sure ds is not averaged before running this cell)

```python
# compare raw RV results to 4 degree averaged RV

RV = ds.RV850.sel(exp='ref').mean('month')
dRV =  ds.RV850.sel(exp='cnt-ref').mean('month')

# plot raw, messy RV data
fig,axs = plt.subplots(1,2,figsize=(12,4),constrained_layout=True)
RV.plot(ax=axs[0], robust=True), dRV.plot(ax=axs[1], robust=True)
fig.suptitle('before averaging')
plt.show()

# 17 x 13 grid points is 4x4 degrees
print(f'\ndlat (17 points): {ds.lat.diff('lat')[0].data*17:.2f}, dlon (13 points): {ds.lon.diff('lon')[0].data*13:.2f}')

# compute 4 degree running mean (cyclical in lon)
yw = 17
xw = 13
RVavg = RV.pad(lon=xw//2, mode='wrap').rolling(lon=xw, lat=yw, min_periods=1).mean().isel(lon=slice(xw//2,-xw//2)).where(RV.notnull())
dRVavg = dRV.pad(lon=xw//2, mode='wrap').rolling(lon=xw, lat=yw, min_periods=1).mean().isel(lon=slice(xw//2,-xw//2)).where(dRV.notnull())

# plot averaged RV data
fig,axs = plt.subplots(1,2,figsize=(12,4), constrained_layout=True)
RVavg.plot(ax=axs[0], robust=True), dRVavg.plot(ax=axs[1], robust=True)
fig.suptitle('with 4 degree averaging')
```
